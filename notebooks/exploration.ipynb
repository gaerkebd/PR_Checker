```python
# Load processed data
with open('../data/processed/prs_processed.json') as f:
    records = json.load(f)

print(f'{len(records)} records')
print(json.dumps(records[0], indent=2)[:1500])
```

```python
# Compare baseline vs RAG for a single PR
with open('../data/processed/baseline_results.json') as f:
    baseline = {r['pr_id']: r for r in json.load(f)}
with open('../data/processed/rag_results.json') as f:
    rag = {r['pr_id']: r for r in json.load(f)}

pr_id = list(baseline.keys())[0]
print('=== BASELINE ===')
print(json.dumps(baseline[pr_id]['review'], indent=2))
print('\n=== RAG ===')
print(json.dumps(rag[pr_id]['review'], indent=2))
```

```python
# Load and display evaluation results
with open('../data/processed/evaluation_results.json') as f:
    eval_data = json.load(f)

print('=== Summary ===')
print(json.dumps(eval_data['summary'], indent=2))
```
